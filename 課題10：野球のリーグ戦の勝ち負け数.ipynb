```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
    
    def calc_win_rate(self):
        return self.win / (self.win + self.lose)
    
    def show_team_result(self):
        rate = self.calc_win_rate()
        print(f"{self.name:9}{self.win:5}{self.lose:5}{self.draw:5} {rate:.3f}")

    @classmethod
    def description(cls):
        print("team       win lose draw rate")
```

```python
Baseball_Team.description()

teames = [
    Baseball_Team(name = "Archers", win = 77, lose = 64, draw = 2),
    Baseball_Team(name = "Meteors", win = 71, lose = 69, draw = 3),
    Baseball_Team(name = "Cheetahs", win = 69, lose = 68, draw = 6),
    Baseball_Team(name = "Sharks", win = 70, lose = 70, draw = 3),
    Baseball_Team(name = "Pythons", win = 68, lose = 73, draw = 2),
    Baseball_Team(name = "Owls", win = 59, lose = 82, draw = 2)
]

for team in teames:
 team.show_team_result()
```
